```python
import torch
from google.colab import userdata
from huggingface_hub import login
from transformers import pipeline
from diffusers import DiffusionPipeline
from datasets import load_dataset
import soundfile as sf
from IPython.display import Audio
```

```python
hf_token=userdata.get('HF_TOKEN')
login(token=hf_token,add_to_git_credential=True)
```

```python
sentiment_analyzer=pipeline(task='sentiment-analysis',device='cuda')
result=sentiment_analyzer('I love this!')
print(result)
```

```python
result=sentiment_analyzer('I should be more excite to be on the way to LLM mastery!!')
print(result)
```

```python
better_stnetiment=pipeline(task='sentiment-analysis',model='nlptown/bert-base-multilingual-uncased-sentiment',device='cuda')
result=better_stnetiment("I should be more excite to be on the way to LLM mastery!!")
print(result)
```

```python
ner=pipeline(task='ner',device='cuda')
result=ner("Ai engineer are learning about the amazing pipelines from HuggingFace in Google colab from Ed Donner")
for i in result:
  print(i)
  print(i['word'],i['entity'])
```

```python
question ="who loves rayhan?"
context="Rayhan is loved by Alisha"
question_answering=pipeline(task="question-answering",device='cuda')
result=question_answering(question=question,context=context)
print(result)
```
